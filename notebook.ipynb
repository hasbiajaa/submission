## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Berapakah presentase penurunan rata-rata jumlah penyewaan harian pada hari dengan cuaca yang buruk dibandingkan cuaca cerah sepanjang 2012, untuk menentukan hari apa yang optimal untuk pemeliharaan armada sepeda yang disewakan.

### Cleaning Data

#### Fixing ... problem

```python
# ==========================================
# 1. Memperbaiki tipe data tanggal
# ==========================================
# Mengubah kolom 'dteday' dari object (string) menjadi datetime
day_df['dteday'] = pd.to_datetime(day_df['dteday'])


# ==========================================
# 2. Memperjelas nilai kategorikal
# ==========================================
# Melakukan mapping nilai angka menjadi label teks yang lebih deskriptif
day_df['season'] = day_df['season'].map({1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'})
day_df['yr'] = day_df['yr'].map({0: '2011', 1: '2012'})
day_df['weathersit'] = day_df['weathersit'].map({
    1: 'Clear/Partly Cloudy',
    2: 'Misty/Cloudy',
    3: 'Light Snow/Rain',
    4: 'Heavy Rain/Snow'
})


# ==========================================
# 3. Mengubah tipe data menjadi category
# ==========================================
# Mengubah kolom-kolom klasifikasi menjadi tipe data 'category'
kolom_kategori = ['season', 'yr', 'mnth', 'holiday', 'weekday', 'workingday', 'weathersit']
for kolom in kolom_kategori:
    day_df[kolom] = day_df[kolom].astype('category')


# ==========================================
# PENGECEKAN HASIL CLEANING
# ==========================================
print("=== INFO DATAFRAME SETELAH CLEANING ===")
day_df.info()

print("\n=== 5 BARIS PERTAMA DATA TERBARU ===")
display(day_df.head())
```

## Exploratory Data Analysis (EDA)

### Explore ...

```python
# ==========================================
# EKSPLORASI DATA UMUM
# ==========================================
# Melihat rangkuman parameter statistik dari data yang sudah bersih
print("=== RANGKUMAN STATISTIK DATA ===")
display(day_df.describe(include="all"))
# ==========================================
# EDA PERTANYAAN BISNIS 1 (Cuaca & Maintenance 2012)
# ==========================================
# 1. Filter data khusus untuk tahun 2012 (kolom 'yr')
data_2012 = day_df[day_df['yr'] == '2012']

# 2. Grouping berdasarkan kondisi cuaca (kolom 'weathersit') dan hitung rata-rata penyewaan (kolom 'cnt')
rata_rata_cuaca = data_2012.groupby(by="weathersit", observed=True).agg({
    "cnt": "mean"
}).reset_index()

# 3. Urutkan dari rata-rata penyewaan tertinggi ke terendah
rata_rata_cuaca = rata_rata_cuaca.sort_values(by="cnt", ascending=False)

print("\n=== RATA-RATA PENYEWAAN SEPEDA BERDASARKAN CUACA (TAHUN 2012) ===")
display(rata_rata_cuaca)
# ==========================================
# EDA PERTANYAAN BISNIS 2 (Fokus Hari Kerja vs Libur di Musim Gugur 2012)
# ==========================================
# 1. Filter data khusus untuk tahun 2012 dan musim gugur (Fall)
data_fall_2012 = day_df[(day_df['yr'] == '2012') & (day_df['season'] == 'Fall')]

# 2. Grouping berdasarkan hari kerja (workingday) dan hitung metrik dari kolom 'cnt'
rata_rata_hari = data_fall_2012.groupby(by="workingday", observed=True).agg({
    "cnt": ["mean", "max", "min", "count"]
}).reset_index()

# 3. Merapikan nama kolom output tabel agar mudah dibaca
rata_rata_hari.columns = ['workingday', 'rata_rata_sewa', 'sewa_tertinggi', 'sewa_terendah', 'jumlah_hari']

# 4. Mapping label workingday biar nggak cuma angka 0 dan 1
rata_rata_hari['workingday'] = rata_rata_hari['workingday'].map({0: 'Akhir Pekan / Libur', 1: 'Hari Kerja'})

print("\n=== ANALISIS PENYEWAAN HARI KERJA VS LIBUR (MUSIM GUGUR 2012) ===")
display(rata_rata_hari)
```

**Insight:** (Opsional)
- Akhir Pekan / Hari Libur: Tercatat ada 29 hari libur/akhir pekan selama musim gugur 2012. Pada hari-hari ini, rata-rata penyewaan sepeda mencapai 6418 sepeda per hari. Menariknya, rekor penyewaan tertinggi harian juga terjadi di akhir pekan, yaitu menembus 8.714 peminjaman.
- Hari Kerja: Terdapat 65 hari kerja. Meskipun jumlah harinya jauh lebih banyak, rata-rata penyewaan hariannya justru lebih rendah, yaitu di angka 7005 sepeda per hari. Angka penyewaan tertinggi di hari kerja adalah 8.173 peminjaman.
- tim maintenance bisa menyimpulkan bahwa hari dengan cuaca buruk adalah waktu yang paling optimal untuk maintenance armada. (pertanyaan bisnis 1)
- Tim operasional harus memprioritaskan jadwal distribusi ulang (rebalancing) stok sepeda dan memperbanyak shift staf lapangan pada akhir pekan selama musim gugur untuk mengantisipasi lonjakan permintaan dari pelanggan kasual maupun pelanggan yang sedang berwisata/bersantai. (pertanyaan bisnis 2)

### Pertanyaan 2:

```python
# ==========================================
# VISUALISASI PERTANYAAN BISNIS 2
# ==========================================
# Data preparation untuk grafik 2 (Melt casual vs registered)
data_fall_melted = data_fall_2012.melt(
    id_vars=['workingday'],
    value_vars=['casual', 'registered'],
    var_name='user_type',
    value_name='count'
)

# Mengubah label workingday di data_fall_melted agar rapi
data_fall_melted['workingday'] = data_fall_melted['workingday'].map({0: 'Akhir Pekan/Libur', 1: 'Hari Kerja'})

# Ubah jadi 2 baris (nrows=2) dan 1 kolom (ncols=1), tinggi kanvas dibesarkan jadi 12
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 12))

# Grafik 1: Bar Chart Keseluruhan - Posisi Atas (ax[0])
sns.barplot(
    x="workingday",
    y="rata_rata_sewa",
    data=rata_rata_hari,
    palette=["#3498db", "#9b59b6"],
    hue="workingday",
    legend=False,
    ax=ax[0]
)
ax[0].set_title("Rata-rata Total Sewa: Hari Kerja vs Libur (Musim Gugur 2012)", fontsize=16, fontweight='bold')
ax[0].set_ylabel("Rata-rata Sewa Harian", fontsize=12)
ax[0].set_xlabel("Kategori Hari", fontsize=12)

# Grafik 2: Bar Chart Breakdown (Casual vs Registered) - Posisi Bawah (ax[1])
sns.barplot(
    x="workingday",
    y="count",
    hue="user_type",
    data=data_fall_melted,
    palette=["#e67e22", "#2c3e50"],
    errorbar=None,
    ax=ax[1]
)
ax[1].set_title("Komposisi Tipe Pengguna: Hari Kerja vs Libur (Musim Gugur 2012)", fontsize=16, fontweight='bold')
ax[1].set_ylabel("Rata-rata Sewa Harian", fontsize=12)
ax[1].set_xlabel("Kategori Hari", fontsize=12)
ax[1].legend(title="Tipe Pengguna", labels=['Kasual', 'Terdaftar (Registered)'], fontsize=11, title_fontsize=12)

plt.tight_layout()
plt.show()
```
